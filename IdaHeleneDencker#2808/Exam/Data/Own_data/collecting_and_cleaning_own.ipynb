# fMRI data

```python
# Define the path to the log files
log_path = '/work/IdaHeleneDencker#2808/Exam/Data/Own_data/fmri'

# Use glob to match all CSV files in the directory
log_files = glob.glob(os.path.join(log_path, "*.csv"))

# Initialize an empty list to store dataframes
df_load = []

# Iterate through sorted log files
for lf in np.sort(log_files):
    print(lf)
    
    # Extract subject ID (e.g., 136) and session number (e.g., 1) using string manipulation
    base_name = os.path.basename(lf)  # e.g., '136_sess_1 (2024-09-24 10-40-28).csv'
    sub_ID = base_name.split('_')[0]  # '136'
    # Read the CSV file
    df_temp = pd.read_csv(lf)
    # Add new columns for subject ID and session number
    df_temp['sub_ID'] = sub_ID
    # Append the dataframe to the list
    df_load.append(df_temp)

# Concatenate all loaded dataframes into one
df = pd.concat(df_load, ignore_index=True)

#drop columns
df.drop(['frames_before_decks',
         'init_mark',
         'frames_after_decks',
         'dur_decks',
         'rt_select',
         'exp_type',
         'dur_win',
         'onset_win',
         'offset_win',
         'condition',
         'offset_decks',
         't_select',
         'ID',
         'onset_decks',
         'n_moves',
         't_moves',
         'dur_measured_decks',
        ], axis=1, inplace=True)


#ADD column of datatype
df['data_type']= 'fmri'

#ADD new column of 'total_pr_trial', so for a row if win=100 and loss=-250, total_pr_trial=-150
df['total_pr_trial'] = df['win'] + df['loss']

#rename total to 'accumulated_score'
df.rename(columns={'total': 'accumulated_score'}, inplace=True)

#rearrange 
df = df[['win', 'loss','total_pr_trial', 'accumulated_score', 'deck_select', 'no', 'session','sub_ID','data_type']]

#new columns 
df['Study'] = 'own'
df['Group'] = 1

# Replace NA in deck_select (currently float64)  with 0 and make int64 
df['deck_select'] = df['deck_select'].fillna(0).astype('int64')

#save dataframe
df.to_csv('/work/IdaHeleneDencker#2808/Exam/Data/Own_data/fmri.csv')
```

```python
#fmri: /fMRI_data (959739)/iowa_exp_script/iowa_exp_data/ (for FMRI)
#MEG: /fMRI_data (959739)/iowa_exp_script/iowa_exp_data/MEG


# Define the path to the log files
log_path = '/work/IdaHeleneDencker#2808/Exam/Data/Own_data/MEG'

# Use glob to match all CSV files in the directory
log_files = glob.glob(os.path.join(log_path, "*.csv"))

# Initialize an empty list to store dataframes
df_load = []

# Iterate through sorted log files
for lf in np.sort(log_files):
    print(lf)
    
    # Extract subject ID (e.g., 136) and session number (e.g., 1) using string manipulation
    base_name = os.path.basename(lf)  # e.g., '136_sess_1 (2024-09-24 10-40-28).csv'
    sub_ID = base_name.split('_')[0]  # '136'
    # Read the CSV file
    df_temp = pd.read_csv(lf)
    # Add new columns for subject ID and session number
    df_temp['sub_ID'] = sub_ID
    # Append the dataframe to the list
    df_load.append(df_temp)

# Concatenate all loaded dataframes into one
df_2 = pd.concat(df_load, ignore_index=True)

#drop columns
df_2.drop(['frames_before_decks',
         'init_mark',
         'frames_after_decks',
         'dur_decks',
         'rt_select',
         'exp_type',
         'dur_win',
         'onset_win',
         'offset_win',
         'condition',
         'offset_decks',
         't_select',
         'ID',
         'onset_decks',
         'n_moves',
         't_moves',
         'dur_measured_decks',
        ], axis=1, inplace=True)


#ADD column of datatype
df_2['data_type']= 'MEG'

#ADD new column of 'total_pr_trial', so for a row if win=100 and loss=-250, total_pr_trial=-150
df_2['total_pr_trial'] = df_2['win'] + df_2['loss']

#rename total to 'accumulated_score'
df_2.rename(columns={'total': 'accumulated_score'}, inplace=True)

#rearrange 
df_2 = df_2[['win', 'loss','total_pr_trial', 'accumulated_score', 'deck_select', 'no', 'session','sub_ID','data_type']]

#new columns 
df_2['Study'] = 'own'
df_2['Group'] = 1

# Replace NA in deck_select (currently float64)  with 0 and make int64 
df_2['deck_select'] = df_2['deck_select'].fillna(0).astype('int64')


#Remove all rows where column 'sub_ID' contains the 'B'
df_2 = df_2[~df_2['sub_ID'].str.contains('B', case=False, na=False)]

#remove the 'A' in column 'sub_ID'
df_2['sub_ID'] = df_2['sub_ID'].str.replace('A', '', case=False, regex=True)


#save dataframe
df_2.to_csv('/work/IdaHeleneDencker#2808/Exam/Data/Own_data/MEG.csv')
```
