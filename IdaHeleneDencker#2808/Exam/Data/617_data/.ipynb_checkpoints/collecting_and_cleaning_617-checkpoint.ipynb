```python
choice = pd.read_csv('/work/IdaHeleneDencker#2808/Exam/Data/617_data/choice_100.csv')
choice

choiceflip =choice.transpose()

choiceflip.reset_index(inplace=True)
choiceflip.rename(columns={'index': 'trial'}, inplace=True)

# Melt the DataFrame to get the desired format
reshaped_df_choice = choiceflip.melt(id_vars=['trial'], var_name='sub_ID', value_name='deck_select')

# Extract the numeric part of ID
reshaped_df_choice['sub_ID'] = reshaped_df_choice['sub_ID'].str.extract('(\d+)').astype(int)

# Extract the numeric part of trial
reshaped_df_choice['trial_Num'] = reshaped_df_choice['trial'].str.extract('(\d+)').astype(int)

# Sort by ID and numeric part of trial
reshaped_df_choice.sort_values(by=['sub_ID', 'trial_Num'], inplace=True)

# Drop the temporary 'trial_Num' column
reshaped_df_choice.drop(columns=['trial_Num'], inplace=True)

# Reset the index for cleaner output
reshaped_df_choice.reset_index(drop=True, inplace=True)

#rename
reshaped_df_choice = reshaped_df_choice.rename(columns={'trial': 'no'})

#cleaning no
reshaped_df_choice['no'] = reshaped_df_choice['no'].str.replace('Choice_', '', case=False, regex=True)


# Display the reshaped DataFrame
reshaped_df_choice
```

```python
win = pd.read_csv('/work/IdaHeleneDencker#2808/Exam/Data/617_data/wi_100.csv')
win

winflip =win.transpose()

winflip.reset_index(inplace=True)
winflip.rename(columns={'index': 'trial'}, inplace=True)

# Melt the DataFrame to get the desired format
reshaped_df_win = winflip.melt(id_vars=['trial'], var_name='sub_ID', value_name='win')

# Extract the numeric part of ID
reshaped_df_win['sub_ID'] = reshaped_df_win['sub_ID'].str.extract('(\d+)').astype(int)

# Extract the numeric part of trial
reshaped_df_win['trial_Num'] = reshaped_df_win['trial'].str.extract('(\d+)').astype(int)

# Sort by ID and numeric part of trial
reshaped_df_win.sort_values(by=['sub_ID', 'trial_Num'], inplace=True)

# Drop the temporary 'trial_Num' column
reshaped_df_win.drop(columns=['trial_Num'], inplace=True) 

# Reset the index for cleaner output
reshaped_df_win.reset_index(drop=True, inplace=True)

#rename
reshaped_df_win = reshaped_df_win.rename(columns={'trial': 'no'})

#cleaning no
reshaped_df_win['no'] = reshaped_df_win['no'].str.replace('Wins_', '', case=False, regex=True)

# Display the reshaped DataFrame
reshaped_df_win
```

```python
loss = pd.read_csv('/work/IdaHeleneDencker#2808/Exam/Data/617_data/lo_100.csv')
loss

lossflip =loss.transpose()

lossflip.reset_index(inplace=True)
lossflip.rename(columns={'index': 'trial'}, inplace=True)

# Melt the DataFrame to get the desired format
reshaped_df_loss = lossflip.melt(id_vars=['trial'], var_name='sub_ID', value_name='loss')

# Extract the numeric part of ID
reshaped_df_loss['sub_ID'] = reshaped_df_loss['sub_ID'].str.extract('(\d+)').astype(int)

# Extract the numeric part of trial
reshaped_df_loss['trial_Num'] = reshaped_df_loss['trial'].str.extract('(\d+)').astype(int)

# Sort by ID and numeric part of trial
reshaped_df_loss.sort_values(by=['sub_ID', 'trial_Num'], inplace=True)

# Drop the temporary 'trial_Num' column
reshaped_df_loss.drop(columns=['trial_Num'], inplace=True) 

# Reset the index for cleaner output
reshaped_df_loss.reset_index(drop=True, inplace=True)

#rename
reshaped_df_loss = reshaped_df_loss.rename(columns={'trial': 'no'})

#cleaning no
reshaped_df_loss['no'] = reshaped_df_loss['no'].str.replace('Losses_', '', case=False, regex=True)

# Display the reshaped DataFrame
reshaped_df_loss


```

```python
df_merged = pd.merge(reshaped_df_win, reshaped_df_loss, on=['no', 'sub_ID'])
df_merged_final = pd.merge(df_merged, reshaped_df_choice, on=['no', 'sub_ID'])


# Merge with index df
df_merged_final_2 = pd.merge(df_merged_final, index, left_on='sub_ID', right_on='Subj')

# Drop the redundant 'Subj' column if needed
df_merged_final_2.drop(columns='Subj', inplace=True)

#add columns
    # total_pr_trial
df_merged_final_2['total_pr_trial'] = df_merged_final_2['win'] + df_merged_final_2['loss']

    # accumulated_score
    # Function to calculate accumulated score for each group
def calculate_accumulated_score(group):
    # Initialize the first accumulated score as total_pr_trial + 2000
    group['accumulated_score'] = 0
    group.loc[group.index[0], 'accumulated_score'] = group['total_pr_trial'].iloc[0] + 2000
    
    # Calculate accumulated score for the rest of the rows
    for i in range(1, len(group)):
        group.loc[group.index[i], 'accumulated_score'] = (
            group['accumulated_score'].iloc[i - 1] + group['total_pr_trial'].iloc[i]
        )
    
    return group

# Group by 'sub_ID' and apply the function
df_merged_final_2 = df_merged_final_2.groupby('sub_ID', group_keys=False).apply(calculate_accumulated_score)


df_merged_final_2['data_type']= 'unknown'

df_merged_final_2['session']= 1
df_merged_final_2['Group']= 2

#rearrange
df_merged_final_2 = df_merged_final_2[['win', 'loss','total_pr_trial', 'accumulated_score', 'deck_select', 'no', 'session','sub_ID','data_type','Study','Group']]

#no as int64
df_merged_final_2['no'] = df_merged_final_2['no'].astype('int64')

#Drop studies i wont need
#those with payoff scheme 2, is most similar to out data: that is Horstmann (n=162), Steingroever (n=70)
#will use Steingroever
final= df_merged_final_2[df_merged_final_2['Study'] == 'SteingroverInPrep']

final.to_csv('/work/IdaHeleneDencker#2808/Exam/Data/617_data/Steingrover_data.csv')

```
